## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.


# 1. Create a copy of the original dataset
df = df.copy()

# 2. Remove unnecessary leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# 3. Convert Date column into datetime format
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%b-%y',
    errors='coerce'
)

# 4. Convert stock price columns into numeric format
price_columns = ['Open', 'High', 'Low', 'Close']

for column in price_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors='coerce'
    )

# 5. Remove rows with missing values
df.dropna(inplace=True)

# 6. Remove duplicate records
df.drop_duplicates(inplace=True)

# 7. Sort dataset according to date
df = df.sort_values(
    by='Date'
).reset_index(drop=True)

# 8. Extract Year and Month from Date
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# 9. Create additional useful features
df['Price_Range'] = df['High'] - df['Low']

df['Price_Change'] = df['Close'] - df['Open']

df['Price_Change_Percentage'] = (
    (df['Close'] - df['Open']) /
    df['Open']
) * 100

# 10. Check dataset after data wrangling
print("Data Wrangling Completed Successfully!")
print("=" * 50)

print("\nDataset Shape:", df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nFinal Dataset:")
display(df.head())

print("\nStatistical Summary:")
display(df.describe().round(2))
````

````python
# Correlation Heatmap visualization code

# Select numerical variables
correlation_data = df[
    ['Open', 'High', 'Low', 'Close',
     'Year', 'Month', 'Price_Range',
     'Price_Change']
]

# Calculate correlation matrix
correlation_matrix = correlation_data.corr()

# Plot heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    square=True,
    cbar=True
)

plt.title(
    'Correlation Heatmap of Yes Bank Stock Variables',
    fontsize=16
)

plt.tight_layout()
plt.show()
````
